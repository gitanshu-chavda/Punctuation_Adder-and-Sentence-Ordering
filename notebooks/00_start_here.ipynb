## 2. Verify NULL markers are preserved (raw vs model_input)

The dataset has 67 literal `NULL` tokens total. Prepared inputs must keep them.

```python
def count_null_in_dir(folder):
    total = 0
    per = {}
    for f in sorted(Path(folder).glob('*.txt')):
        t = f.read_text(encoding='utf-8')
        c = t.count('NULL')
        per[f.name] = c
        total += c
    return total, per

raw_total, raw_per = count_null_in_dir('data/raw/samples')
inp_total, inp_per = count_null_in_dir('data/model_input')
print('Raw total NULL:', raw_total)
print('Model input total NULL:', inp_total)
print('Match:', raw_total == inp_total)
for k in raw_per:
    print(f"  {k}: raw={raw_per[k]} input={inp_per.get(k, 'NA')}")
```
