# 02 — NULL Marker Diagnostic

The dataset contains **67 literal `NULL` tokens** with **unknown meaning**. This notebook compares raw and punctuated text around each occurrence (in order) to see how punctuation restoration interacts with `NULL`.

**Do not** delete or reinterpret `NULL` based on this alone.

```python
import os, sys
from pathlib import Path
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)
print('CWD:', Path.cwd())
```

```python
# Run the CLI diagnostic script logic inline for notebook viewing
from scripts.null_diagnostic import analyze_file
from pathlib import Path

files = sorted(p.name for p in Path('data/raw/samples').glob('*.txt'))
summary = []
for name in files:
    r = analyze_file(name)
    summary.append(r)
    print('='*80)
    print(f"{name}: raw={r['raw_count']} punct={r['punct_count']}")
    for occ in r['occurrences']:
        print(f"  #{occ['n']:02d} right raw->punct: {occ['raw_right']!r} -> {occ['punct_right']!r} | left: {occ['raw_left']!r} -> {occ['punct_left']!r}")
        print(f"       PUNC: ...{occ['punct_ctx']}...")
    print()

print('TOTALS:')
print('  raw =', sum(x['raw_count'] for x in summary))
print('  punct =', sum(x['punct_count'] for x in summary))
```

## How to interpret

- **Counts match (67/67)**: literal `NULL` strings were preserved in outputs.
- **`punct_right` becomes `:` or `।` etc.**: the model inserted punctuation adjacent to `NULL` in the output text. This is an observation, not proof of what `NULL` means.
- Compare informational samples (1-5) vs dialogue-like samples (6-7) for different patterns.

If you see a count mismatch, stop and investigate file alignment/encoding issues before drawing conclusions.
